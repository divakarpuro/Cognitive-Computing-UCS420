Q1. Consider the following sensor readings:
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
A sensor has an average measurement error of +2°C.
Perform the following without using a for loop:

a. Add 2°C to every reading using vectorization.

b. Convert all temperatures from Celsius to Fahrenheit using:

F = 9/5 C + 32

c. Identify all readings greater than 32°C using Boolean indexing.

d. Count how many readings exceed 32°C.

e. Explain how vectorization and Boolean indexing can make data processing more efficient 
than explicitly iterating through every value

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

temperature2 = temperature + 2

print(temperature2)

fahrenheit = (9 / 5) * temperature2 + 32

print(fahrenheit)

readings_greater_32 = temperature2[temperature2 > 32]

print(readings_greater_32)

count = np.sum(temperature2 > 32)

print(count)

# Vectorization allows NumPy to perform operations on all elements 
# of an array at onece without directly using a for loop. 
# It makes the code shorter and faster because 
# NumPy performs the operations efficiently at the array level.

# Boolean indexing allows us to select elements that 
# satisfy a given condition without manually checking each value. 
#Thats why vectorization and Boolean indexing make data processing
# more efficient and convenient than explicitly iterating through every value.
```

Q2. The following matrix represents the number of daily steps recorded by four users 
over three days:
steps = np.array([
[5000, 6200, 7100],
[8000, 7500, 9000],
[4500, 5100, 4800],
[9000, 8500, 9500]])

Perform the following:

a. Calculate the total steps recorded.

b. Calculate the mean number of steps.

c. Find the maximum and minimum values.

d. Calculate the total steps for each day using axis=0.

e. Calculate the total steps for each user using axis=1.

f. Find the user/day position containing the maximum number of steps using argmax().

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

total_steps = np.sum(steps)

print(total_steps)

mean_steps = np.mean(steps)

print(mean_steps)

maximum = np.max(steps)

print(maximum)

minimum = np.min(steps)

print(minimum)

total_dayWise = np.sum(steps, axis=0)

print(total_dayWise)

total_userWise = np.sum(steps, axis=1)

print(total_userWise)


max_index = np.argmax(steps)
position = np.unravel_index(max_index, steps.shape)

print("flatten index:", max_index)
print("position:", position)
print("max steps:", steps[position])
```

Q4. A simple cognitive-assistive system wants to predict a user's assistance score based 
on three factors:
• Sleep hours 
• Activity level 
• Stress level 
Consider:

X = np.array([
[6, 70, 3],
[5, 50, 6],
[8, 80, 2],
[4, 30, 8]
])

y = np.array([40, 65, 30, 85])

Here, each row represents one user and each column represents a feature.
Perform the following:

a. Display the shape and dimensions of X.

b. Calculate: X.T and explain what the transpose represents.

c. Calculate the matrix product: X.T @ X

d. Calculate the matrix inverse using: np.linalg.inv() where appropriate.

f. Explain what the resulting coefficients represent in the context of the three user

g. A new user has: new_user = np.array([5, 40, 7]) Use your calculated coefficients to 
obtain a predicted assistance score

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
], dtype=float)

y = np.array([40, 65, 30, 85], dtype=float)

print("shape:", X.shape)
print("dimension:", X.ndim)


XT = X.T
print("transpose:", XT)


XTX = XT @ X
print("X.T @ X:", XTX)


XTX_inv = np.linalg.inv(XTX)
print("inverse:", XTX_inv)


beta = XTX_inv @ XT @ y
print("coeff:", beta)


print("sleep coeff:", beta[0])
print("activity coeff:", beta[1])
print("stress coeff:", beta[2])


new_user = np.array([5, 40, 7], dtype=float)

predicted_score = new_user @ beta

print("predicted score:", predicted_score)
```

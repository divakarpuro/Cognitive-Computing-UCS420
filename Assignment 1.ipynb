```python
def add_odd(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i

    return total


n = int(input("Enter the value of n: "))
print("Sum of odd numbers =", add_odd(n))
```

WAP using function that add all prime numbers from 1 to n, n given by the user.

```python
def add_prime(n):
    total = 0

    for num in range(2, n + 1):
        is_prime = True

        for i in range(2, num//2 + 1):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            total += num

    return total


n = int(input("Enter the value of n: "))
print("Sum of prime numbers =", add_prime(n))
```

```python
import sys

sys.argv = ["Program.py", "Divakar", "Purohit"]

print("Arguments:", sys.argv)

s = sys.argv[1] + " " + sys.argv[2]

print(sys.argv[1], "+", sys.argv[2], "-->", s)
```

15.3 Add all numbers

```python
import sys

sys.argv = ["Program.py", "1", "7", "6", "6"]

print("Arguments:", sys.argv)

total = 0

for x in sys.argv[1:]:
    total += int(x)

print("Sum is -->", total)
```

16 File Handling

```python
16.1 Writing 1 to 10 in file
```

```python
fp=open('CHECK1.txt','w')	
for  i in range(1,11):
	fp.write(str(i) + "\n")	
fp.close()

print ("WRITTEN !! ")
```

16.2 Read a file and print its content

```python
fp=open('CHECK1.txt')		
for line in fp: 		    
	print (line.strip())
fp.close()
```


16.3 Read from one file, Convert it to upper case and write to other file

```python
Readfp=open('CHECK1.txt')		
Writefp=open('CHECK2.txt','w')	
for line in Readfp:
	Writefp.write(line.upper())

Writefp.close()
Readfp.close()

print ("WRITTEN!! \n")
```

Q.2 Questions on Basic NumPy Array: 

a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5]) 

b) Find the most frequent value and their indice(s) in the following arrays: 

i. x = np.array([1,2,3,4,5,1,2,1,1,1]) 

ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ]) 

```python
arr = np.array([1, 2, 3, 6, 4, 5]) 
rev = np.flip(arr)
print(arr)
print(rev)


x = np.array([1,2,3,4,5,1,2,1,1,1]) 
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ]) 

values1, counts1 = np.unique(x, return_counts=True)
max_count1 = np.max(counts1)
most_frequent1 = values1[counts1 == max_count1]
indices1 = np.where(x == most_frequent1[0])[0]

print('Most Frequent value in x : ',most_frequent1)
print('indices : ',indices1)

values2, counts2 = np.unique(y, return_counts=True)
max_count2 = np.max(counts2)
most_frequent2 = values2[counts2 == max_count2]
indices2 = np.where(x == most_frequent2[0])[0]
print('Most Frequent value in y : ',most_frequent2)
print('indices : ',indices2)

```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>> 
with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, 
median, max, min, unique elements. Reshape the array to four rows and three columns and 
name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three 
columns and name it as resized_ ucs420_<your_name>>.

```python
ucs420_Divakar = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print(ucs420_Divakar)

print("mean:", np.mean(ucs420_Divakar))
print("median:", np.median(ucs420_Divakar))
print("maximum:", np.max(ucs420_Divakar))
print("minimum:", np.min(ucs420_Divakar))
print("unique elements:", np.unique(ucs420_Divakar))

reshaped_ucs420_Divakar = ucs420_Divakar.reshape(4, 3)

print(reshaped_ucs420_Divakar)

resized_ucs420_Divakar = np.resize(ucs420_Divakar, (2, 3))

print(resized_ucs420_Divakar)
```
